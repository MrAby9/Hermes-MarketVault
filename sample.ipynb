```python
import os
import psycopg2
from dotenv import load_dotenv

# 1. Carica le variabili dal file .env
load_dotenv()

# 2. Connettiti al database Neon
conn = psycopg2.connect(
    host=os.getenv("PGHOST"),
    database=os.getenv("PGDATABASE"),
    user=os.getenv("PGUSER"),
    password=os.getenv("PGPASSWORD"),
    sslmode=os.getenv("PGSSLMODE")
)
cur = conn.cursor()

# 3. Crea una tabella di prova
query = """
CREATE TABLE IF NOT EXISTS tabella_prova (
    id SERIAL PRIMARY KEY,
    nome VARCHAR(50),
    data_creazione DATE DEFAULT CURRENT_DATE
);
"""
cur.execute(query)
conn.commit() # FONDAMENTALE: salva le modifiche sul DB

print("Tabella creata con successo nel database Neon!")

cur.close()
conn.close()
```
